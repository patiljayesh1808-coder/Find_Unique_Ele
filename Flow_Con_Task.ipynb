#### Task 1) 

- WAP to get a unique element list form above list without changing data position.

```python
l = [5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]
unique_ele = []

for data in l:
    if data not in unique_ele:
        unique_ele.append(data)
print(unique_ele)
```

#### Task 2)

- find the frequency of every element from above input list in dictionary.

```python
l = [5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]

freq_list = {}

for data in l:
    if data in freq_list:
        freq_list[data] = freq_list[data] + 1

    else:
        freq_list[data] = 1

print(f"How many times repeated each element: {freq_list}")
```
